19. Display the images (Original, Lower Contrast, Higher Contrast).

```python
matrix1 = np.ones(bgr_img.shape) * 0.8
matrix2 = np.ones(bgr_img.shape) * 1.2
img_lower  = np.uint8(cv2.multiply(np.float64(bgr_img), matrix1))
img_higher = np.uint8(cv2.multiply(np.float64(bgr_img), matrix2))
plt.figure(figsize = [18,5])
plt.subplot(131); plt.imshow(img_lower[:, :, ::-1]);  plt.title('Lower Contrast')
plt.subplot(132); plt.imshow(bgr_img[:, :, ::-1]);        plt.title('Original')
plt.subplot(133); plt.imshow(img_higher[:, :, ::-1]); plt.title('Higher Contrast');
```

20. Split the image (boy.jpg) into the B,G,R components & Display the channels.

```python
by_img = cv2.imread('boy.jpg', cv2.IMREAD_COLOR)
b, g, r = cv2.split(by_img)
plt.figure(figsize = [20, 10])
plt.subplot(141); plt.imshow(b); plt.title('Blue Channel')
plt.subplot(142); plt.imshow(g); plt.title('Green Channel')
plt.subplot(143); plt.imshow(r); plt.title('red Channel')
imgMerged = cv2.merge((b, g, r))
plt.subplot(144)
plt.imshow(imgMerged)
plt.title('Merged Output');
```
